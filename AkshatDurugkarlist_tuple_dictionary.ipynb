```python
# Write a program
#    Original list : [(10, 20, 30), (40, 50, 60), (70, 80, 90)]
#    Expected output : [(10, 20, 100), (40, 50, 100), (70, 80, 100)]

list1 = [(10, 20, 30), (40, 50, 60), (70, 80, 90)]
new_list = []

for tuple1 in list1:
    new_tuple = tuple1[:-1] + (100,)
    new_list.append(new_tuple)

print("Original list :", list1)
print("Updated list :", new_list)
```

```python
# Python program to find the maximum and minimum element in a tuple.

numbers = (10, 20, 30, 40, 50)
print(max(numbers))
print(min(numbers))
```

```python
# Python program to swap the first and last element of a tuple.

numbers = (10, 20, 30, 40, 50)

new_tuple = (numbers[:-1],) + (numbers[1:1],) = (numbers[0],)

print("Original :", numbers)
print("Updated :", new_tuple)
```

```python
# Write a program to concatenate the following dictionaries to create a new one.
#    Sample Dictionary : dictionary1 = {1:10, 2:20}, dictionary2 = {3:30, 4:40}, dictionary3 = {5:50, 6:60}
#    Expected Result : {1:10, 2:20, 3:30, 4:40, 5:50, 6:60}

dictionary1 = {1:10, 2:20}
dictionary2 = {3:30, 4:40}
dictionary3 = {5:50, 6:60}

print("Dictionary1 :", dictionary1)
print("Dictionary2 :", dictionary2)
print("Dictionary3 :", dictionary3)

new_dictionary = {}

new_dictionary.update(dictionary1)
new_dictionary.update(dictionary2)
new_dictionary.update(dictionary3)

print("New dictionary :", new_dictionary)
```

```python
# Write a program to iterate over a dictionary using for loop and print the keys alone, values alone and both keys and values.

mydictionary = {
    101:"Amit",
    102:"Rahul",
    103:"Priya",
}

print("Keys :")
for key in mydictionary:
    print(key)

print("\nValues :")
for value in mydictionary.values():
    print(value)

print("\nKeys and Values :")
for key, value in mydictionary.items():
    print(key, value)
```

```python
# Convert list to dictionary.
#    Example : list1 = ['name', 'panel', 'rollno'], list2 = ['ABC', 'B', 34]
#              mydictionary = {'name':'ABC', 'panel':'B', 'rollno':34}

list1 = ['name', 'panel', 'rollno']
list2 = ['ABC', 'B', 34]

mydictionary = dict(zip(list1, list2))

print("Dictionary :", mydictionary)
```

```python
# Find the mean of values of dictionary.
#    Example : mydict = {'marks1':23, 'marks2':123, 'marks3':43, 'marks4':13, 'marks5':39}
#    output : mean of all values.

mydictionary = {'marks1':23, 'marks2':123, 'marks3':43, 'marks4':13, 'marks5':39}
total = sum(mydictionary.values())
mean = total / len(mydictionary)
print("Mean of all values :", mean)
```

```python
# write a python program to print the highest marks in a dictionary

mydictionary = {'marks1':10, 'marks2':20, 'marks3':30, 'marks4':40}
hmarks = max(mydictionary, key=mydictionary.get())
print(hmarks)
```

```python
# python program to count the frequency of each item in a list.

numbers = [10, 20, 30, 40, 10, 20, 30]
frequency = {}
for num in numbers:
    if num in frequency:
      frequency[num] = frequency[num] + 1
    else:
      frequency[num] = 1
print ('Frequency of elements', frequency)
```

```python
# python program to sort a dictionary by its values

my_dict = {'marks1':20, 'marks2':10, 'marks3':30, 'marks4':25}
sorted_dict = dict(sorted(my_dict.items(), key=lambda item: item[1]))
print('Dictionary sorted by values:', sorted_dict)
```

```python
my_dict = {'marks1':20, 'marks2':10, 'marks3':30, 'marks4':25}
sorted_dict = dict(sorted(my_dict.items(), key=lambda item: item[1], reverse = True))
print('Dictionary sorted by values:', sorted_dict)
```
