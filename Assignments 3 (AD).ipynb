```python
aa = input ('Enter the length of perpendicular')
bb = input ('Enter the length of base')
cc = input ('Enter the length of hypotenuse')
a = float (aa)
b = float (bb)
c = float (cc)
if a+b < c :
    print ('Not a right angled triangle')
elif a**2 + b**2 == c**2 :
    print ('Is a right angled triangle')
else :
    print ('Not a right angled triangle')
```
