```python
x = ['a', 'b', 'c', 'Akshat', True, [3, 4, 5]]
print(x[:3]) # starting index number isnt mentioned (starts from the first element)
print(x[2:]) # last index number isnt mentioned (ends at the last element)
print(x[1:4]) # prints from 1st to 3rd index number
print(x[:]) # prints all the elements
print(x[::-1]) # reversing 
print(x[::2]) # skipping every second element
print(x[::3]) # 
print(x[3::2]) 
```

```python
x = [1, 2, 3, 4, 5]
y = [5, 6, 7, 8, 9, 10]
for a in x:
    for b in y:
        if a==b:
            print ('atleast one common element')
            break
        elif a!=b:
            print ('no common element')
            break
```
