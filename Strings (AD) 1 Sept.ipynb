```python
a = 'Hello World!'
print(a) # strings are written in single quotes (') or double quotes (")
```

```python
a = 'Hello World!'
b = a + ' Hello Again'
print (b) # string concatenation
```

```python
a = 'Hello World!' # converts all the upper case letters to lower case letters
a.lower() # using string library
```

```python
a = 'Hello World!' # converts all the lower case letters to upper case letters 
a.upper()
```

```python
a = 'Hello World!' # splits a sentence into words
a.split()
```

```python
a = 'Hello World!'
a.split('e')
```

```python
a = 'Hello World!' # string slicing 
print (a[1:4])
```

```python
a = 'Hello World!'
print (a[::2]) # skips 1 place
```

```python
a = 'Hello World!'
print (a[::-1]) # reverses string
```

```python
a = 'Hello World!'
print (a[4]) # prints the element (letter) at index number 4
```

```python
a = 'Hello World!'
b = len (a) # number of strings or length of string
print (b) 
```
