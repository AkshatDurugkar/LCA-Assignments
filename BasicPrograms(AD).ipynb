```python
aa = input ('enter 1st number')
bb = input ('enter 2nd number')
op = input ('enter operation 1=+, 2=-, 3=*, 4=/')
a = float(aa)
b = float(bb)
o = float(op)
if o == 1:
    print (a+b)
elif o == 2:
    print (a-b)
elif o == 3:
    print (a*b)
elif o == 4:
    if b != 0:
        print (a/b)
    else :
        print ('denominator can't be zero')
else :
    print ('invalid operation')
```
